```python
import pandas as pd
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt

# Load dataset
df = pd.read_csv("IPL_Data.csv")

# Select only numerical columns
X = df.select_dtypes(include=['int64', 'float64'])

print(X.columns)

# Remove missing values
X = X.dropna()

# K-Means
kmeans = KMeans(n_clusters=4, random_state=42)
df_cluster = X.copy()
df_cluster["Cluster"] = kmeans.fit_predict(X)

# Plot using first two numeric columns
plt.figure(figsize=(10,6))

plt.scatter(
    df_cluster.iloc[:,0],
    df_cluster.iloc[:,1],
    c=df_cluster["Cluster"],
    cmap="viridis",
    s=50
)

plt.scatter(
    kmeans.cluster_centers_[:,0],
    kmeans.cluster_centers_[:,1],
    c="red",
    marker="X",
    s=250,
    label="Centroids"
)

plt.xlabel(X.columns[0])
plt.ylabel(X.columns[1])
plt.title("IPL Player Clustering using K-Means")
plt.legend()

plt.show()
```

```python
# Select required columns
X = df[['RunsScored', 'BattingAVG', 'BattingS/R', 'Wickets', 'EconomyRate']]

# Remove rows with missing values
df_clean = df.dropna(subset=['RunsScored', 'BattingAVG', 'BattingS/R', 'Wickets', 'EconomyRate'])

# Features
X = df_clean[['RunsScored', 'BattingAVG', 'BattingS/R', 'Wickets', 'EconomyRate']]

# KMeans
from sklearn.cluster import KMeans

kmeans = KMeans(n_clusters=3, random_state=42)

df_clean['Cluster'] = kmeans.fit_predict(X)

# Display result
df_clean[['Name', 'Cluster']].head()
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10,6))

plt.scatter(
    df_clean['BattingAVG'],
    df_clean['RunsScored'],
    c=df_clean['Cluster'],
    cmap='viridis',
    s=80
)

plt.xlabel("Batting Average")
plt.ylabel("Runs Scored")
plt.title("IPL Player Segmentation using K-Means")

plt.show()
```
